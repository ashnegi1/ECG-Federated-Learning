# Signal processing demonstration
Visualize a raw lead and the exact configured Butterworth + DWT output. Dataset access is via WFDB.

```python
import matplotlib.pyplot as plt
import wfdb
from ecg_fl.preprocessing.dsp import preprocess_ecg
record = wfdb.rdrecord('../data/raw/mitdb/100', physical=True)
raw = record.p_signal[:3600, 0]
clean = preprocess_ecg(raw, record.fs)
plt.figure(figsize=(12,4)); plt.plot(raw, label='raw', alpha=.6); plt.plot(clean, label='Butterworth 0.5–45 Hz + DWT'); plt.legend(); plt.show()
```
