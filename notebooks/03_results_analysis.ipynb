# Results analysis
Reads metrics generated by real experiment runs; no metrics are embedded in this notebook.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt
paths = [Path('../outputs/experiment/centralized_metrics.json'), Path('../outputs/experiment/federated_metrics.json')]
for p in paths:
    if p.exists():
        m=json.loads(p.read_text()); print(p.name, {k:m[k] for k in ('accuracy','f1_macro','f1_weighted')})
        plt.figure(); plt.imshow(m['confusion_matrix']); plt.title(p.stem); plt.xlabel('predicted'); plt.ylabel('true'); plt.show()
    else: print('Run the experiment to create', p)
```
